```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

##This cell imports all the required libraries for data manipulation, visualization, and machine learning preparation. Display settings are adjusted to show all dataset columns.

```python
df = pd.read_csv('/content/drive/MyDrive/DATA SET /cleaned_merged_dataset_v3.csv')

df.head()
```

##The final cleaned master dataset is loaded successfully. This dataset will be used for feature selection and machine learning model preparation.

```python
date_columns = [
    'booking_date',
    'ship_date',
    'expected_delivery_date',
    'actual_delivery_date',
    'registration_date',
    'forecast_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])

print(df[date_columns].dtypes)

df['shipment_delay'] = (
    df['actual_delivery_date'] -
    df['expected_delivery_date']
).dt.days

df['delay_days'] = df['shipment_delay'].clip(lower=0)

df['shipment_delayed'] = (
    df['shipment_delay'] > 0
).astype(int)

df[['shipment_delay','delay_days','shipment_delayed']].head()
```

##Shipment delay was calculated using the difference between actual and expected delivery dates. A binary target variable named shipment_delayed was created where 0 indicates on-time delivery and 1 indicates delayed delivery.

```python
# Booking Features
df['booking_year'] = df['booking_date'].dt.year
df['booking_month'] = df['booking_date'].dt.month
df['booking_day'] = df['booking_date'].dt.day
df['booking_weekday'] = df['booking_date'].dt.dayofweek

# Shipping Features
df['ship_year'] = df['ship_date'].dt.year
df['ship_month'] = df['ship_date'].dt.month
df['ship_day'] = df['ship_date'].dt.day
df['ship_weekday'] = df['ship_date'].dt.dayofweek

# Expected Delivery Features
df['expected_year'] = df['expected_delivery_date'].dt.year
df['expected_month'] = df['expected_delivery_date'].dt.month
df['expected_day'] = df['expected_delivery_date'].dt.day
df['expected_weekday'] = df['expected_delivery_date'].dt.dayofweek

# Forecast Features
df['forecast_month'] = df['forecast_date'].dt.month
df['forecast_weekday'] = df['forecast_date'].dt.dayofweek

# Customer Registration Features
df['registration_year'] = df['registration_date'].dt.year
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

categorical_cols = df.select_dtypes(include='object').columns

print("Categorical Columns")

print(categorical_cols)

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])

print("\nEncoding Completed Successfully")
```

##Interpretation

All categorical variables were converted into numerical values using Label Encoding. This preprocessing step enables machine learning algorithms to process categorical information effectively.

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x='shipment_delayed',
    palette='Set2'
)

plt.title("Class Distribution of Shipment Delay")

plt.xlabel("Shipment Status")

plt.ylabel("Number of Shipments")

plt.xticks(
    [0, 1],
    ["On-Time", "Delayed"]
)

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

##Interpretation

The bar chart visually represents the class distribution of the target variable. It shows that on-time shipments are more frequent than delayed shipments, indicating a moderate class imbalance. Understanding this distribution is important because an imbalanced dataset can influence model performance and evaluation.
