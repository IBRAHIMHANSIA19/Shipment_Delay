```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)
```

##Interpretation

This cell imports the necessary Python libraries required for performing univariate analysis. Pandas and NumPy are used for data manipulation, while Matplotlib and Seaborn are used to create statistical visualizations. The display option is configured to show all dataset columns for easier analysis.

```python
date_columns = [
    'booking_date',
    'ship_date',
    'expected_delivery_date',
    'registration_date',
    'actual_delivery_date',
    'forecast_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])

df.info()
```

##Interpretation

This cell provides a summary of the dataset, including the total number of records, number of columns, data types, non-null values, and memory usage. It confirms that the dataset is complete and ready for univariate analysis.

```python
print(df.columns.tolist())
```

```python
categorical_cols = [
    'shipping_mode',
    'shipment_type',
    'priority',
    'carrier_name',
    'carrier_type',
    'customer_type',
    'industry',
    'customs_required',
    'documentation_complete',
    'inspection_required',
    'cargo_type',
    'category',
    'hazardous',
    'perishable',
    'temperature_controlled',
    'route_risk',
    'delivery_status',
    'vehicle_type',
    'fuel_type',
    'maintenance_status',
    'warehouse_type',
    'weather_condition'
]
```

```python
numerical_cols = [
    'weight_kg',
    'volume_cbm',
    'declared_value',
    'average_rating',
    'fleet_size',
    'years_of_service',
    'weight_per_unit',
    'distance_km',
    'average_transit_days',
    'traffic_index',
    'delivery_attempts',
    'capacity_kg',
    'vehicle_age',
    'warehouse_capacity',
    'current_utilization',
    'temperature',
    'rainfall',
    'humidity',
    'wind_speed',
    'visibility'
]
```

```python
numerical_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()

# Remove old delay column
if 'shipment_delay' in numerical_cols:
    numerical_cols.remove('shipment_delay')

print(numerical_cols)
```

##Interpretation

This cell separates the dataset into numerical and categorical features based on their data types. This classification allows suitable statistical methods and visualizations to be applied to each type of variable.

```python
df[numerical_cols].describe().T
```

##Interpretation

This cell generates descriptive statistics for all numerical features, including count, mean, standard deviation, minimum, quartiles, and maximum values. These statistics provide an overview of the central tendency and variability of the numerical data.

```python
summary = pd.DataFrame({
    'Mean': df[numerical_cols].mean(),
    'Median': df[numerical_cols].median(),
    'Mode': df[numerical_cols].mode().iloc[0],
    'Standard Deviation': df[numerical_cols].std(),
    'Minimum': df[numerical_cols].min(),
    'Maximum': df[numerical_cols].max(),
    'Skewness': df[numerical_cols].skew()
})

summary
```

##Interpretation

This table summarizes the key statistical measures for each numerical feature. It provides information about the average value, central tendency, variability, range, and skewness of the data. Skewness indicates whether the distribution is symmetric, positively skewed, or negatively skewed.

```python
for col in numerical_cols:

    plt.figure(figsize=(7,4))

    sns.histplot(df[col], kde=True)

    plt.title(f'Distribution Plot of {col}')

    plt.xlabel(col)

    plt.ylabel("Frequency")

    plt.show()
```

##Interpretation

This cell creates histogram and Kernel Density Estimation (KDE) plots for each numerical feature. These visualizations help understand the distribution, concentration, spread, and skewness of the data.

```python
for col in numerical_cols:

    plt.figure(figsize=(7,2))

    sns.boxplot(x=df[col])

    plt.title(f'Box Plot of {col}')

    plt.show()
```

##Interpretation

This cell generates box plots for every numerical feature. Box plots provide a graphical summary of the data distribution and help identify potential outliers, quartiles, and the median value.

```python
for col in categorical_cols:

    plt.figure(figsize=(10,5))

    sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index
    )

    plt.title(f'Count Plot of {col}')
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)

    plt.show()
```

##Interpretation

This cell creates count plots for each categorical variable. These visualizations display the frequency of each category and make it easier to compare category distributions visually.

```python
df.columns
```

```python
# KDE Plot for Numerical Columns

for col in numerical_cols:
    plt.figure(figsize=(8, 5))
    sns.kdeplot(data=df, x=col, fill=True)
    plt.title(f'KDE Plot of {col}')
    plt.xlabel(col)
    plt.ylabel('Density')
    plt.grid(True)
    plt.show()
```

##Interpretation

This code generates a Kernel Density Estimation (KDE) plot for each numerical feature in the dataset.
A KDE plot provides a smooth estimate of the data distribution, making it easier to understand the underlying pattern compared to a histogram.
These plots help identify whether the data is normally distributed, positively skewed, negatively skewed, or contains multiple peaks (multimodal distribution).
The distribution of each numerical feature can be further analyzed to determine whether transformations or preprocessing steps are required before model development.

```python
for col in categorical_cols:
    print("="*60)
    print("Column :", col)

    print("\nMost Frequent Category")
    print(df[col].value_counts().head(1))

    print("\nLeast Frequent Category")
    print(df[col].value_counts().tail(1))
```
