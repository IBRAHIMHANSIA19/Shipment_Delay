```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv('/content/drive/MyDrive/DATA SET /carriers.csv')
df2 = pd.read_csv('/content/drive/MyDrive/DATA SET /customers.csv')
df3 = pd.read_csv('/content/drive/MyDrive/DATA SET /customs.csv')
df4 = pd.read_csv('/content/drive/MyDrive/DATA SET /products.csv')
df5 = pd.read_csv('/content/drive/MyDrive/DATA SET /routes.csv')
df6 = pd.read_csv('/content/drive/MyDrive/DATA SET /shipment_delivery_history.csv')
df7 = pd.read_csv('/content/drive/MyDrive/DATA SET /shipments.csv')
df8 = pd.read_csv('/content/drive/MyDrive/DATA SET /vehicles.csv')
df9 = pd.read_csv('/content/drive/MyDrive/DATA SET /warehouses.csv')
df10 = pd.read_csv('/content/drive/MyDrive/DATA SET /weather.csv')
```

##Interpretation

This cell imports the required Python libraries, Pandas and NumPy, for data manipulation and analysis. It then loads the ten logistics-related CSV files into separate DataFrames (df1 to df10). Each dataset represents a different operational aspect of the logistics system, such as carriers, customers, products, routes, vehicles, warehouses, customs, weather conditions, shipment history, and shipments. Loading these datasets is the first step in preparing the data for integration and further preprocessing.

```python
# Merge all datasets into one Master Dataset

master_df = (
    df7
    .merge(df1, on='carrier_id', how='inner')
    .merge(df2, on='customer_id', how='inner')
    .merge(df3, on='customs_id', how='inner')
    .merge(df4, on='product_id', how='inner')
    .merge(df5, on='route_id', how='inner')
    .merge(df6, on='shipment_id', how='inner')
    .merge(df8, on='vehicle_id', how='inner')
    .merge(df9, on='warehouse_id', how='inner')
    .merge(df10, on='weather_id', how='inner')
)

print("Master Dataset Shape:", master_df.shape)

master_df.head()
```

##Interpretation

This cell combines all ten datasets into a single integrated dataset named master_df using Left Join operations. The shipments dataset (df7) serves as the primary table because it contains the foreign keys required to connect with all other datasets. Each related dataset is merged using its corresponding key (such as carrier_id, customer_id, product_id, and route_id). Using a Left Join ensures that all shipment records are preserved even if matching information is unavailable in the related tables. The resulting master dataset provides a comprehensive view of shipment operations and serves as the primary dataset for preprocessing, exploratory data analysis, feature engineering, and machine learning.

##Interpretation

This cell calculates the percentage of missing values for each column in the merged dataset and displays only those columns that contain missing data in descending order. Evaluating the percentage of missing values helps determine the most suitable preprocessing strategy. Columns with a low percentage of missing values can generally be imputed using statistical methods such as the median or mode, while columns with a high percentage of missing values may require more careful consideration. In this dataset, all missing-value percentages are relatively low, making imputation an appropriate approach without removing any columns or records.

```python
# ================================
# Fill Missing Values
# ================================

# Numerical Columns (Fill with Median)
numerical_cols = [
    'fleet_size',
    'declared_value',
    'weight_per_unit',
    'traffic_index',
    'delivery_attempts',
    'current_utilization',
    'humidity',
    'wind_speed',
    'volume_cbm'
]

for col in numerical_cols:
    if col in df.columns:
        df[col].fillna(df[col].median(), inplace=True)

# Categorical Columns (Fill with Mode)
categorical_cols = [
    'priority',
    'customer_type',
    'industry',
    'state_x',
    'documentation_complete',
    'inspection_required',
    'hs_code',
    'supplier_name',
    'route_risk',
    'history_id',
    'delivery_status',
    'fuel_type',
    'maintenance_status',
    'visibility'
]

for col in categorical_cols:
    if col in df.columns:
        df[col].fillna(df[col].mode()[0], inplace=True)

# Date Columns (Fill with Most Frequent Date)
date_cols = [
    'registration_date',
    'actual_delivery_date'
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')
        df[col].fillna(df[col].mode()[0], inplace=True)

print("✅ All missing values have been handled successfully.")
```

##Interpretation

This cell handles all missing values in the merged dataset using appropriate imputation techniques based on the data type of each feature. Missing values in numerical columns are replaced with the median, as it is less affected by outliers and preserves the overall distribution of the data. Missing values in categorical columns are filled with the mode (most frequently occurring value), ensuring data consistency while maintaining the most common category. Date columns are first converted into the datetime format and then any missing dates are replaced with the most frequently occurring date. This preprocessing step ensures that the dataset is complete and suitable for further analysis and machine learning without losing valuable records.

```python
# ============================================
# Feature Engineering
# ============================================

# Calculate Shipment Delay (in days)

df['shipment_delay'] = (
    df['actual_delivery_date'] -
    df['expected_delivery_date']
).dt.days

df[['actual_delivery_date',
    'expected_delivery_date',
    'shipment_delay']].head()
```

##Interpretation

This cell calculates the shipment delay by computing the difference between the actual delivery date and the expected delivery date. Positive values indicate late deliveries, negative values indicate early deliveries, and zero represents on-time deliveries. This feature forms the basis for shipment delay analysis.

```python
# Create Delay Days
# Negative values become zero

df['delay_days'] = df['shipment_delay'].clip(lower=0)

df[['shipment_delay','delay_days']].head()
```

##Interpretation

This cell creates a new feature called delay_days, which represents only the number of delayed days. Early deliveries are converted to zero because they do not contribute to shipment delays. This feature provides a more meaningful business measure for analyzing late deliveries.

```python
# Create Binary Target Variable

df['shipment_delayed'] = (df['delay_days'] > 0).astype(int)

df[['delay_days','shipment_delayed']].head()
```

##Interpretation

This cell creates the binary target variable shipment_delayed. A value of 1 indicates that a shipment was delivered late, while 0 represents shipments delivered on time or earlier. This target variable is appropriate for building a shipment delay prediction model using classification algorithms.

```python
print(df[['shipment_delay',
          'delay_days',
          'shipment_delayed']].describe())

print("\n")

print(df['shipment_delayed'].value_counts())
```
