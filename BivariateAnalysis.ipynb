```python
date_columns = [
    'booking_date',
    'ship_date',
    'expected_delivery_date',
    'registration_date',
    'actual_delivery_date',
    'forecast_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])
```

##Interpretation

This cell converts all date-related columns into datetime format, enabling accurate calculations involving shipment and delivery dates.

```python
# Convert date columns
df['expected_delivery_date'] = pd.to_datetime(df['expected_delivery_date'])
df['actual_delivery_date'] = pd.to_datetime(df['actual_delivery_date'])

# Shipment Delay (can be negative if delivered early)
df['shipment_delay'] = (
    df['actual_delivery_date'] -
    df['expected_delivery_date']
).dt.days

# Delay Days (only positive delays)
df['delay_days'] = df['shipment_delay'].clip(lower=0)

# Binary Target
df['shipment_delayed'] = (df['shipment_delay'] > 0).astype(int)

# Verify
df[['shipment_delay','delay_days','shipment_delayed']].head()
```

##Interpretation

Scatter plots illustrate the relationship between important numerical variables and shipment delay. These visualizations help identify trends, clusters, and potential correlations between shipment characteristics and delivery performance.

```python
categorical_features = [
    'shipping_mode',
    'priority',
    'vehicle_type',
    'weather_condition',
    'route_risk'
]

for col in categorical_features:

    plt.figure(figsize=(10,5))

    sns.boxplot(
        data=df,
        x=col,
        y='delay_days'
    )

    plt.xticks(rotation=45)

    plt.title(f'{col} vs Delay Days')

    plt.xlabel(col)

    plt.ylabel("Delay Days")

    plt.grid(axis='y', linestyle='--', alpha=0.5)

    plt.show()
```

##Interpretation

Box plots compare the distribution of shipment delays across different categories. They help determine whether certain shipping modes, vehicle types, weather conditions, or route risks are associated with higher delays.

```python
for col in categorical_features:

    plt.figure(figsize=(10,5))

    (
        df.groupby(col)['delay_days']
          .mean()
          .sort_values(ascending=False)
          .plot(kind='bar', color='steelblue', edgecolor='black')
    )

    plt.ylabel("Average Delay Days")

    plt.xlabel(col)

    plt.title(f'Average Delay Days by {col}')

    plt.xticks(rotation=45)

    plt.grid(axis='y', linestyle='--', alpha=0.5)

    plt.show()
```

##Interpretation

These bar charts display the average shipment delay for each category. They provide a clear comparison of operational factors influencing delivery performance.

```python
cross_tab = pd.crosstab(
    df['priority'],
    df['shipment_delayed']
)

cross_tab
```

##Interpretation

This cross-tabulation summarizes the relationship between shipment priority and delivery status, helping identify whether higher-priority shipments tend to have better delivery performance.

```python
cross_tab.plot(
    kind='bar',
    figsize=(8,5),
    edgecolor='black'
)

plt.title("Shipment Priority vs Shipment Delay Status")

plt.xlabel("Shipment Priority")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=0)

plt.legend(
    title="Shipment Delayed",
    labels=["No (0)", "Yes (1)"]
)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
