```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

pd.set_option('display.max_columns', None)
```

##Interpretation

This cell imports all the required Python libraries for data preprocessing, dataset splitting, and Logistic Regression model development.

```python
df = pd.read_csv('/content/drive/MyDrive/DATA SET /cleaned_merged_dataset_v3.csv')

df.head()
```

##Interpretation

The final cleaned shipment dataset is loaded successfully. This dataset will be used to build the baseline Logistic Regression model.

```python
date_columns = [

    'booking_date',

    'ship_date',

    'expected_delivery_date',

    'actual_delivery_date',

    'registration_date',

    'forecast_date'

]

for col in date_columns:

    df[col] = pd.to_datetime(df[col])

print("Date Columns Converted Successfully")
```

##Interpretation

All date-related columns are converted into DateTime format so that meaningful time-based features can be extracted during feature engineering.

```python
df['shipment_delay'] = (

    df['actual_delivery_date'] -

    df['expected_delivery_date']

).dt.days

df['delay_days'] = df['shipment_delay'].clip(lower=0)

df['shipment_delayed'] = (

    df['shipment_delay'] > 0

).astype(int)

df[['shipment_delay','delay_days','shipment_delayed']].head()
```

##Interpretation

Shipment delay is calculated using the difference between the actual and expected delivery dates. A binary target variable named shipment_delayed is created where 0 represents on-time deliveries and 1 represents delayed shipments.

```python
# Booking Date
df['booking_year'] = df['booking_date'].dt.year
df['booking_month'] = df['booking_date'].dt.month
df['booking_day'] = df['booking_date'].dt.day
df['booking_weekday'] = df['booking_date'].dt.dayofweek

# Ship Date
df['ship_year'] = df['ship_date'].dt.year
df['ship_month'] = df['ship_date'].dt.month
df['ship_day'] = df['ship_date'].dt.day
df['ship_weekday'] = df['ship_date'].dt.dayofweek

# Expected Delivery
df['expected_year'] = df['expected_delivery_date'].dt.year
df['expected_month'] = df['expected_delivery_date'].dt.month
df['expected_day'] = df['expected_delivery_date'].dt.day
df['expected_weekday'] = df['expected_delivery_date'].dt.dayofweek

# Registration
df['registration_year'] = df['registration_date'].dt.year

# Forecast
df['forecast_month'] = df['forecast_date'].dt.month
df['forecast_weekday'] = df['forecast_date'].dt.dayofweek

print("Feature Engineering Completed Successfully")
```

##Interpretation

Date-based features are extracted from the original date columns to help the Logistic Regression model identify seasonal, monthly, and weekly shipment patterns that may influence shipment delays.

```python
free_text_columns = [
    'customer_name',
    'carrier_name',
    'supplier_name',
    'product_name',
    'origin_city',
    'destination_city',
    'origin_port',
    'destination_port',
    'vehicle_number',
    'warehouse_name'
]

df.drop(
    columns=free_text_columns,
    errors='ignore',
    inplace=True
)

print("Free Text Columns Removed Successfully")
```

##Interpretation

Free-text columns are removed because they contain descriptive information that cannot be directly processed by most machine learning algorithms.

```python
label_encoder = LabelEncoder()

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])

print("Categorical Features Encoded Successfully")

print("\nRemaining Object Columns")

print(df.select_dtypes(include='object').columns)
```

##Interpretation

All categorical variables are converted into numerical values using Label Encoding. This transformation allows machine learning algorithms to process categorical information efficiently.

```python
# Remove unnecessary columns

drop_columns = [

    # Target
    'shipment_delayed',

    # Target Leakage
    'shipment_delay',
    'delay_days',
    'actual_delivery_date',

    # ID Columns
    'shipment_id',
    'booking_id',
    'customer_id',
    'product_id',
    'carrier_id',
    'warehouse_id',
    'route_id',
    'vehicle_id',
    'customs_id',
    'weather_id',
    'history_id',

    # Original Date Columns
    'booking_date',
    'ship_date',
    'expected_delivery_date',
    'registration_date',
    'forecast_date'
]

X = df.drop(columns=drop_columns)

y = df['shipment_delayed']

print("Feature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)
```

##Interpretation

The dataset is divided into the feature matrix (X) and the target variable (y). Identifier columns, original date columns, and target leakage variables are removed because they do not contribute to prediction or may provide information that would unfairly influence the model. The remaining engineered features are retained for machine learning.

```python
print("Feature Matrix Shape :", X.shape)

print("Target Shape :", y.shape)

print("\nFeature Data Types\n")

print(X.dtypes.value_counts())

print("\nRemaining Object Columns")

print(X.select_dtypes(include='object').columns)

print("\nRemaining Datetime Columns")

print(X.select_dtypes(include='datetime64[ns]').columns)
```

##Interpretation

The feature matrix and target variable are verified before model training. The data types are checked to ensure that all features are numerical and compatible with the Logistic Regression algorithm. The absence of object and datetime columns confirms that the dataset is fully prepared for model training.

```python
print("="*80)

print("Final Features Selected for Model Training")

print("="*80)

for i, col in enumerate(X.columns, start=1):
    print(f"{i}. {col}")

print("\nTotal Features :", len(X.columns))
```

##Interpretation

The final list of selected features is displayed to document the input variables that will be used by the Logistic Regression model. This improves transparency and reproducibility.

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("Dataset Split Successfully")

print("Training Feature Matrix :", X_train.shape)

print("Testing Feature Matrix :", X_test.shape)

print()

print("Training Target :", y_train.shape)

print("Testing Target :", y_test.shape)
```

##Interpretation

The dataset is divided into training and testing sets using an 80:20 ratio. The training set is used to build the Logistic Regression model, while the testing set is reserved for evaluating its performance. Stratified sampling ensures that the class distribution is preserved in both subsets.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_columns = X_train.select_dtypes(
    include=['int64', 'float64', 'int32']
).columns

X_train[numeric_columns] = scaler.fit_transform(
    X_train[numeric_columns]
)

X_test[numeric_columns] = scaler.transform(
    X_test[numeric_columns]
)

print("Numerical Features Scaled Successfully")
```

##Interpretation

Numerical features are standardized using StandardScaler so that they have a mean of zero and a standard deviation of one. Feature scaling improves the convergence and stability of the Logistic Regression algorithm.

```python
from sklearn.linear_model import LogisticRegression
```

##Interpretation

The Logistic Regression algorithm is imported from Scikit-learn. It will be used as the first baseline classification model for predicting whether a shipment will be delayed.

```python
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print(model)
```

##Interpretation

A Logistic Regression model is initialized with a maximum of 1000 iterations to ensure proper convergence during training. The random_state parameter is set to 42 to make the training process reproducible.

```python
model.fit(X_train, y_train)

```

```python
print("Model Verification")

print("-" * 50)

print("Model Type :", type(model).__name__)

print("Training Completed :", hasattr(model, "coef_"))

print("Number of Features Used :", X_train.shape[1])

print("Number of Classes :", len(model.classes_))

print("Classes :", model.classes_)
```

```python
pred_train = model.predict(X_train)

pred_test = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, mean_absolute_percentage_error,mean_absolute_percentage_error,mean_squared_error,r2_score
```

```python
print('Train Accuracy Score:-\n', accuracy_score(y_train,pred_train))
print('Train Confusion Matrix:-\n', confusion_matrix(y_train,pred_train))
print('Train Mean Squared Error:-\n', mean_squared_error(y_train,pred_train))
print('Train R2 Score:-\n', r2_score(y_train,pred_train))
```

```python
print('Test Accuracy Score:-\n', accuracy_score(y_test,pred_test))
print('Test Confusion Matrix:-\n', confusion_matrix(y_test,pred_test))
print('Test Mean Squared Error:-\n', mean_squared_error(y_test,pred_test))
print('Test R2 Score:-\n', r2_score(y_test,pred_test))
```

##Interpretation

This cell verifies that the Logistic Regression model has been successfully trained. The presence of learned coefficients confirms that the model fitting process completed successfully and that the model is ready for evaluation.

```python
feature_coefficients = pd.DataFrame({

    "Feature": X_train.columns,

    "Coefficient": model.coef_[0]

})

feature_coefficients = feature_coefficients.sort_values(
    by="Coefficient",
    ascending=False
)

feature_coefficients.head(15)
```

##Interpretation

The Logistic Regression coefficients indicate how each feature influences shipment delay prediction. Features with positive coefficients increase the probability of delay, while features with negative coefficients decrease it.

```python
top_features = feature_coefficients.head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    data=top_features,
    x="Coefficient",
    y="Feature",
    color="steelblue"
)

plt.title("Top 10 Features Influencing Shipment Delay")

plt.xlabel("Coefficient")

plt.ylabel("Feature")

plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()

plt.show()
```

##Interpretation

The bar chart highlights the most influential features identified by the Logistic Regression model. These variables have the strongest positive relationship with shipment delays and provide useful business insights into the factors affecting delivery performance.
