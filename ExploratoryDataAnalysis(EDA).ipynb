```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

##Interpretation

This cell imports the required Python libraries for data manipulation, statistical analysis, and visualization. These libraries are used throughout the exploratory data analysis process to identify trends, patterns, and relationships within the logistics dataset.

```python
df = pd.read_csv('/content/drive/MyDrive/DATA SET /cleaned_merged_dataset_v3.csv')

df.head()
```

##Interpretation

This cell loads the cleaned and merged logistics dataset into a DataFrame. Displaying the dataset dimensions and the first five records confirms that the dataset has been successfully loaded and is ready for exploratory data analysis.

```python
date_columns = [
    'booking_date',
    'ship_date',
    'expected_delivery_date',
    'registration_date',
    'actual_delivery_date',
    'forecast_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])
```

##Interpretation

This cell converts all date-related columns into the datetime format, enabling accurate time-based calculations such as delivery delay analysis.

```python
df['shipment_delay'] = (
    df['actual_delivery_date'] -
    df['expected_delivery_date']
).dt.days

df['delay_days'] = df['shipment_delay'].clip(lower=0)

df['shipment_delayed'] = (
    df['shipment_delay'] > 0
).astype(int)
```

```python
df[['shipment_delay', 'delay_days', 'shipment_delayed']].head()
```

```python
# Check Distribution of Target Variable

print(df['shipment_delayed'].value_counts())

print("\nPercentage Distribution:\n")

print(round(df['shipment_delayed'].value_counts(normalize=True) * 100, 2))
```

##Interpretation

This cell displays the number and percentage of delayed and non-delayed shipments. It helps evaluate the balance of the target variable and determine whether the dataset is suitable for classification modeling or if class imbalance handling techniques may be required.

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x='shipment_delayed',
    palette='Set2'
)

plt.title("Distribution of Delayed and On-Time Shipments", fontsize=14)

plt.xlabel("Shipment Status (0 = On Time/Early, 1 = Delayed)")

plt.ylabel("Number of Shipments")

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

##Interpretation

This visualization illustrates the distribution of delayed and non-delayed shipments in the dataset. It provides a clear overview of the target variable and helps identify whether delayed shipments occur frequently or represent a smaller proportion of the overall shipment records.

```python
carrier_analysis = df['carrier_name'].value_counts()

plt.figure(figsize=(12,5))

carrier_analysis.plot(
    kind='bar',
    color='steelblue',
    edgecolor='black'
)

plt.title("Carrier-wise Shipment Analysis", fontsize=14)

plt.xlabel("Carrier Name")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=45, ha='right')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This visualization compares the number of shipments handled by each logistics carrier. It helps identify the busiest carriers and provides insights into carrier workload and shipment distribution across the logistics network.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='shipping_mode',
    order=df['shipping_mode'].value_counts().index,
    palette='Set2'
)

plt.title("Shipping Mode Comparison", fontsize=14)

plt.xlabel("Shipping Mode")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=30)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This chart compares shipment volumes across different shipping modes. It highlights the most frequently used transportation methods and helps understand customer preferences for shipment services.

```python
# Create Route Name (Run only once)

if 'route_name' not in df.columns:
    df['route_name'] = df['origin_city'] + " → " + df['destination_city']

# Top 15 Most Used Routes

top_routes = df['route_name'].value_counts().head(15)

plt.figure(figsize=(13,6))

top_routes.plot(
    kind='bar',
    color='darkorange',
    edgecolor='black'
)

plt.title("Top 15 Routes by Shipment Volume", fontsize=14)

plt.xlabel("Route")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=60, ha='right')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This visualization displays the top 15 origin-to-destination routes based on shipment volume. It helps identify the busiest transportation corridors, supporting route optimization, capacity planning, and resource allocation within the logistics network.

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='weather_condition',
    hue='shipment_delayed',
    palette='Set2'
)

plt.title("Impact of Weather Condition on Shipment Delays", fontsize=14)

plt.xlabel("Weather Condition")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=30)

plt.legend(
    title="Shipment Delayed",
    labels=["On Time/Early (0)", "Delayed (1)"]
)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This visualization compares delayed and non-delayed shipments across different weather conditions. It helps identify whether adverse weather conditions contribute to higher shipment delays and provides insights into environmental factors affecting logistics performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='distance_km',
    y='delay_days',
    hue='shipment_delayed',
    alpha=0.7
)

plt.title("Distance vs Delay Days", fontsize=14)

plt.xlabel("Distance (km)")

plt.ylabel("Delay Days")

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This scatter plot illustrates the relationship between shipment distance and delivery delay. It helps determine whether longer transportation distances are associated with increased delivery delays and operational challenges.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='weight_kg',
    y='delay_days',
    hue='shipment_delayed',
    alpha=0.7
)

plt.title("Package Weight vs Delay Days", fontsize=14)

plt.xlabel("Package Weight (kg)")

plt.ylabel("Delay Days")

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This visualization analyzes whether package weight has an impact on delivery delay. It helps determine whether heavier shipments are more likely to experience transportation delays.

```python
# Select Numerical Columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Remove the original shipment_delay column
if 'shipment_delay' in numerical_cols:
    numerical_cols.remove('shipment_delay')

plt.figure(figsize=(18,12))

sns.heatmap(
    df[numerical_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features", fontsize=16)

plt.tight_layout()

plt.show()
```

##Interpretation

The correlation heatmap displays the strength and direction of relationships among all numerical variables in the dataset. It helps identify important features that are strongly correlated with shipment delays and other logistics performance metrics.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='priority',
    hue='shipment_delayed',
    palette='Set2'
)

plt.title("Shipment Priority vs Shipment Delay", fontsize=14)

plt.xlabel("Shipment Priority")

plt.ylabel("Number of Shipments")

plt.legend(
    title="Shipment Delayed",
    labels=["On Time/Early (0)", "Delayed (1)"]
)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This chart compares shipment delays across different priority levels. It helps determine whether high-priority shipments are delivered more efficiently than standard or low-priority shipments.

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='vehicle_type',
    hue='shipment_delayed',
    palette='Set2'
)

plt.title("Vehicle Type vs Shipment Delay", fontsize=14)

plt.xlabel("Vehicle Type")

plt.ylabel("Number of Shipments")

plt.xticks(rotation=30)

plt.legend(
    title="Shipment Delayed",
    labels=["On Time/Early (0)", "Delayed (1)"]
)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

##Interpretation

This visualization compares delayed and non-delayed shipments across different vehicle types. It helps identify whether certain vehicle categories are associated with higher delivery delays.

```python
carrier_delay = (
    df.groupby('carrier_name')['delay_days']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(12,5))

carrier_delay.plot(
    kind='bar',
    color='tomato',
    edgecolor='black'
)

plt.title("Average Delay by Carrier")

plt.xlabel("Carrier")

plt.ylabel("Average Delay (Days)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
weather_delay = (
    df.groupby('weather_condition')['delay_days']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

weather_delay.plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.title("Average Delay by Weather Condition")

plt.ylabel("Average Delay (Days)")

plt.tight_layout()

plt.show()
```
